## 1. Importação de Bibliotecas e Setup

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Adiciona o diretório raiz ao path para poder importar módulos de 'src'
sys.path.append(os.path.abspath(".."))

from src.data_loader import load_wine_data
from src.preprocessing import create_target_variable, split_data

# Estilo de visualizações premium
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 120
```

## 2. Carregamento dos Dados (via `kagglehub`)

```python
# Carrega o dataset completo do Kaggle
df_raw = load_wine_data()
print(f"Linhas: {df_raw.shape[0]} | Colunas: {df_raw.shape[1]}")
df_raw.head()
```

## 3. Criação da Variável Alvo Binária

```python
# Binarização: qualidade >= 7 é considerado vinho de alta qualidade (1)
df = create_target_variable(df_raw, source_col="quality", target_col="high_quality")
df.head()
```

## 4. Distribuição das Classes (Alta vs Baixa/Média Qualidade)

```python
# Analisando o desbalanceamento de classes
dist = df["high_quality"].value_counts(normalize=True) * 100
print("Distribuição percentual da variável alvo:")
print(dist)

# Gráfico estilizado com paleta premium
sns.countplot(x="high_quality", data=df, palette=["#8d99ae", "#2b2d42"])
plt.title("Distribuição de Vinhos de Alta Qualidade (Classe 1) vs Demais (Classe 0)", fontsize=14, pad=15, weight="bold")
plt.xlabel("Alta Qualidade (quality >= 7)")
plt.ylabel("Quantidade")
plt.xticks([0, 1], ["Baixa/Média (0)", "Alta (1)"])
plt.show()
```
